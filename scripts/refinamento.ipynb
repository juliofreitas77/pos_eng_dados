```python
import duckdb
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
df = con.execute(""" select * from produtos """).fetchdf()
df.head()
```

```python
df2 = df.drop(columns=['ID_CATEGORIA', 'ID_FORNECEDOR'])
```

```python
df2.head()
con.execute("""
drop table dim_produtos
""")
```

```python
con.execute("""
create table if not exists dim_produtos (
            id_produto bigint,
            nm_produto text,
            vl_produto FLOAT
            )
""")
```

```python
df_dim = con.execute(""" select * from dim_produtos """).fetchdf()
df_dim.head()
```

```python
con.execute("""
    INSERT INTO dim_produtos SELECT * FROM df2
""")
```

```python
con.close()
```
